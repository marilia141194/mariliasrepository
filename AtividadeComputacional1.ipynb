```python
import numpy as np
import matplotlib.pyplot as plt

def funcao_linear(a, b):
    if a == 0:
        print("Valor de a invalido!")
    elif a > 0:
        print("A reta é crescente.")
    else:
        print("A reta é decrescente.")

    if a != 0:
      y_intercept = b
      x_intercept = -b / a
      print(f"A reta cruza o eixo y em: {y_intercept:.4f}")
      print(f"A reta cruza o eixo x em: {x_intercept:.4f}")

      x_min = min(0, x_intercept) - 2
      x_max = max(0, x_intercept) + 2
      x = np.linspace(x_min, x_max, 100)

      y = a * x + b
      plt.plot(x, y, label=f"y = {a}x + {b}")
      plt.axhline(0, color='black', linewidth=1)
      plt.axvline(0, color='black', linewidth=1)
      plt.scatter([x_intercept], [0], color='red')
      plt.scatter([0], [y_intercept], color='blue')
      plt.xlim(x_min, x_max)
      plt.ylim(min(y), max(y_intercept, y_intercept + 2))
      plt.legend()
      plt.show()

def funcao_quadratica(a, b, c):
    if a == 0:
        print("Valor de a invalido!")
    elif a > 0:
        print("A parábola é côncava para cima.")
    else:
        print("A parábola é côncava para baixo.")

    if a != 0 :
      y_intercept = c
      delta = b**2 - 4*a*c
      x_vertice = -b / (2 * a)
      y_vertice = -delta / (4 * a) if delta != 0 else 0

      if delta < 0:
        print("A parábola não cruza o eixo x.")
        x1, x2 = x_vertice - 5, x_vertice + 5
      elif delta == 0:
          x1 = x2 = x_vertice
          print(f"A parábola cruza o eixo x em: {x1:.4f}")
      else:
          x1 = (-b + np.sqrt(delta)) / (2 * a)
          x2 = (-b - np.sqrt(delta)) / (2 * a)
          print(f"A parábola cruza o eixo x em x1: {x1:.4f} e x2: {x2:.4f}")

      print(f"A parábola cruza o eixo y em: {y_intercept:.4f}")
      print(f"Vértice em x = {x_vertice:.4f}, y = {y_vertice:.4f}")

      x = np.linspace(min(x1, x2) - 2, max(x1, x2) + 2, 100)
      y = a * x**2 + b * x + c
      plt.plot(x, y, label=f"y = {a}x² + {b}x + {c}")
      plt.axhline(0, color='black', linewidth=1)
      plt.axvline(0, color='black', linewidth=1)
      plt.scatter([0], [y_intercept], color='blue')
      plt.scatter([x_vertice], [y_vertice], color='green', label="Vértice")
      if delta >= 0:
          plt.scatter([x1, x2], [0, 0], color='red')
      plt.scatter([x_vertice], [y_vertice], color='red')
      plt.ylim(min(y), max(y_intercept, y_vertice) + 2)
      plt.legend()
      plt.show()

def main():
    print("Escolha uma opção:")
    print("1. Função Linear")
    print("2. Função Quadrática")
    opcao = int(input("Digite a opção desejada: "))

    if opcao == 1:
        a = float(input("Digite o coeficiente a: "))
        b = float(input("Digite o coeficiente b: "))
        funcao_linear(a, b)
    elif opcao == 2:
        a = float(input("Digite o coeficiente a: "))
        b = float(input("Digite o coeficiente b: "))
        c = float(input("Digite o coeficiente c: "))
        funcao_quadratica(a, b, c)
    else:
        print("Opção inválida.")

if __name__ == "__main__":
    main()
```
